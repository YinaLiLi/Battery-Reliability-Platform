```python
from scipy.io import loadmat
from pathlib import Path

DATA_DIR = Path("../data/raw/nasa_batteries")
BATTERIES = ["B0005", "B0006", "B0007", "B0018"]

for battery_id in BATTERIES:
    print(f"\n{'='*60}")
    print(f"Battery: {battery_id}")
    print("="*60)

    # Load battery
    mat = loadmat(DATA_DIR / f"{battery_id}.mat")
    battery = mat[battery_id]
    cycles = battery["cycle"][0, 0]

    print(f"Number of cycles: {cycles.shape[1]}")
    print(f"Cycle fields: {cycles.dtype.names}")

    # Count operation types
    operation_counts = {}

    for i in range(cycles.shape[1]):
        cycle = cycles[0, i]
        operation = cycle["type"][0]

        operation_counts[operation] = (
            operation_counts.get(operation, 0) + 1
        )

    print(f"Operation counts: {operation_counts}")

    # Inspect one example of each operation type
    seen = set()

    for i in range(cycles.shape[1]):
        cycle = cycles[0, i]
        operation = cycle["type"][0]

        if operation not in seen:
            seen.add(operation)

            print(f"\nExample {operation} cycle:")
            print(f"Ambient temperature: {cycle['ambient_temperature']}")
            print(f"Data fields: {cycle['data'].dtype.names}")

        if len(seen) == 3:
            break
```
